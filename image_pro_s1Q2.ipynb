```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
```

```python
image=cv2.imread("Lenna_(test_image).png",1)
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

```python
bright_inc=cv2.add(image, np.ones( image.shape, dtype=np.uint8)*50)
```

```python
bright_dec=cv2.subtract(image, np.ones( image.shape, dtype=np.uint8)*50)
```

```python
contrast_inc=cv2.convertScaleAbs(image, alpha=1.5, beta=0)
```

```python
contrast_dec=cv2.convertScaleAbs(image, alpha=0.5, beta=0)
```
