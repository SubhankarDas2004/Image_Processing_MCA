```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = cv2.imread("Lenna_(test_image).png")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')
plt.show()

rows, cols = img.shape[:2]
print(rows,cols)
```

TRANSLATION

```python
M_trans = np.float32([[1, 0, 100], [0, 1, 50]])
translated = cv2.warpAffine(img_rgb, M_trans, (cols, rows))

plt.imshow(translated)
plt.axis('off')
plt.show()
```

SCALING

```python
scaled = cv2.resize(img_rgb, None, fx=1.5, fy=1.5, interpolation=cv2.INTER_LINEAR)

plt.imshow(scaled)
plt.axis('off')
plt.show()
```

ROTATION

```python
M_rot = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1)
rotated = cv2.warpAffine(img_rgb, M_rot, (cols, rows))

plt.imshow(rotated)
plt.axis('off')
plt.show()
```

HORIZONTAL REFLECTION

```python
h_flip = cv2.flip(img_rgb, 1)

plt.imshow(h_flip)
plt.axis('off')
plt.show()
```

VERTICAL REFLECTION

```python
v_flip = cv2.flip(img_rgb, 0)

plt.imshow(v_flip)
plt.axis('off')
plt.show()
```

Q2. Zooming and Shrinking with Interpolation

 2x Zooming

```python
zoom_2_nearest = cv2.resize(img_rgb, None, fx=2, fy=2, interpolation=cv2.INTER_NEAREST)
zoom_2_bilinear = cv2.resize(img_rgb, None, fx=2, fy=2, interpolation=cv2.INTER_LINEAR)
zoom_2_bicubic = cv2.resize(img_rgb, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1); plt.imshow(zoom_2_nearest); plt.title("2x - Nearest"); plt.axis('off')
plt.subplot(1,3,2); plt.imshow(zoom_2_bilinear); plt.title("2x - Bilinear"); plt.axis('off')
plt.subplot(1,3,3); plt.imshow(zoom_2_bicubic); plt.title("2x - Bicubic (Best)"); plt.axis('off')
plt.show()
```

4x Zooming

```python
zoom_4_nearest = cv2.resize(img_rgb, None, fx=4, fy=4, interpolation=cv2.INTER_NEAREST)
zoom_4_bilinear = cv2.resize(img_rgb, None, fx=4, fy=4, interpolation=cv2.INTER_LINEAR)
zoom_4_bicubic = cv2.resize(img_rgb, None, fx=4, fy=4, interpolation=cv2.INTER_CUBIC)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1); plt.imshow(zoom_4_nearest); plt.title("4x - Nearest "); plt.axis('off')
plt.subplot(1,3,2); plt.imshow(zoom_4_bilinear); plt.title("4x - Bilinear"); plt.axis('off')
plt.subplot(1,3,3); plt.imshow(zoom_4_bicubic); plt.title("4x - Bicubic"); plt.axis('off')
plt.show()
```

```python
shrink_half_nearest = cv2.resize(img_rgb, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_NEAREST)
shrink_half_bilinear = cv2.resize(img_rgb, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)
shrink_half_bicubic = cv2.resize(img_rgb, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_CUBIC)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1); plt.imshow(shrink_half_nearest); plt.title("1/2 - Nearest"); plt.axis('off')
plt.subplot(1,3,2); plt.imshow(shrink_half_bilinear); plt.title("1/2 - Bilinear"); plt.axis('off')
plt.subplot(1,3,3); plt.imshow(shrink_half_bicubic); plt.title("1/2 - Bicubic"); plt.axis('off')
plt.show()
```
